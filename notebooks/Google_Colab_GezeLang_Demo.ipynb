**GezeLang** is a high-performance open-source programming language transpiler supporting 15 Ethiopic dialects (Amharic, Ge'ez, Tigrinya, Gurage, Gojjami, Gondari, Wollo, Harari, etc.) with native WebAssembly execution, sub-0.6s batch parsing, bidirectional transpilation (`Python <-> GezeLang`), and interactive Jupyter/Colab kernel integration.

Developed at **Adama Science and Technology University (ASTU)**.

## 1. Install & Load GezeLang Extension
Simply load `%load_ext gezelang` to enable the `%%gezelang` cell magic.

```python
# In Google Colab, uncomment to install:
# !pip install gezeLang
%load_ext gezelang
```

## 2. Hello World in Amharic
Write standard Amharic functions with `ስራ` (def), `መልስ` (return), and `አሳይ` (print).

```python
%%gezelang
ስራ ሰላምታ(ስም):
    መልስ f"ሰላም፣ {ስም}! እንኳን ወደ GezeLang በደህና መጣህ።"

አሳይ(ሰላምታ("አበበ"))
```

## 3. Seamless Python <-> GezeLang Interoperability
Variables and functions defined in `%%gezelang` cells immediately live in your notebook namespace, accessible to standard Python cells and libraries like NumPy or Matplotlib!

```python
# Standard Python cell calling the Amharic function defined above!
message = ሰላምታ("ጫላ")
print(f"[Python cell output]: {message}")
```

## 4. Loops & Computational Logic
Use `ለእያንዳንዱ` (for), `ውስጥ` (in), `እስከሆነ` (while), `ከሆነ` (if), `ካልሆነ` (else).

```python
%%gezelang
ድምር = 0
ለእያንዳንዱ ቁጥር ውስጥ ክልል(1, 11):
    ድምር = ድምር + ቁጥር

አሳይ(f"ከ 1 እስከ 10 ያሉት ቁጥሮች ድምር: {ድምር}")
ከሆነ ድምር > 50:
    አሳይ("ድምሩ ከ 50 ይበልጣል!")
ካልሆነ:
    አሳይ("ድምሩ ከ 50 ያንሳል")
```

## 5. Multi-Dialect Support (Tigrinya & Classical Ge'ez)
Switch dialect with the `--lang` flag or set sticky dialect with `%gezelang_lang`.

```python
%%gezelang --lang tigrinya
ስራ ተደምር(ሀ, ለ):
    መልስ ሀ + ለ

ኣርእይ(f"ትግርኛ ድምር: {ተደምር(50, 75)}")
```

```python
%%gezelang --lang geez
ግበር ተቀበል(ስም):
    አግብእ f"ሰላም ለእከ ኦ {ስም}"

አይድዕ(ተቀበል("ገብረ ማርያም"))
```

## 6. Bidirectional Reverse Transpiler (`Python -> GezeLang`)
Convert existing Python code into Ethiopic script code using `--reverse`!

```python
%%gezelang --reverse
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(10))
```

## 7. Inspecting Transpiled Python Output (`-p` / `--show-python`)
Inspect the generated Python before execution:

```python
%%gezelang --show-python
ስራ ብዜት(ሀ, ለ):
    መልስ ሀ * ለ
አሳይ(ብዜት(6, 7))
```
